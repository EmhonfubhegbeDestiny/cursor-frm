```python
wordsincode = []
class Cursor:
    def cursor():
        add = input('type a charcater')
        wordsincode.append(add)
        wordsincode.append('|')
    def add(hello):
        checkcursor = wordsincode.index('|')
        wordsincode[checkcursor] = hello
        zzz = checkcursor + 1
        wordsincode.insert(zzz,'|')
    def empty():
        for _ in range(len(wordsincode)):
            wordsincode.pop()
    def printl():
        while '|' not in wordsincode:
            wordsincode.append('|')
        printed = ''.join(wordsincode)
        return printed
    def movecursortotheleft():
        yowa = wordsincode.index('|')
        hafana = yowa - 1
        wordsincode[yowa], wordsincode[hafana] = wordsincode[hafana],wordsincode[yowa]
    def movecursortotheright():
        yowa = wordsincode.index('|')
        hafana = yowa + 1
        wordsincode[yowa], wordsincode[hafana] = wordsincode[hafana],wordsincode[yowa]
    def backspace():
        apple = wordsincode.index('|')
        yes = apple - 1 
        if apple == 0:
            return True
        else:
            wordsincode.pop(yes)
            return True

```

```python
hey = Cursor
hey.empty()
```
